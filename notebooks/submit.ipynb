```python
from pathlib import Path

from pyopensky.s3 import S3Client
```

```python
s3 = S3Client()

bucket = "prc-2026-incredible-rose"
file_path = Path("incredible-rose_v2.parquet")
object_name = "incredible-rose_v2.parquet"

result = s3.s3client.fput_object(
    bucket,
    object_name,
    str(file_path),
    content_type="application/vnd.apache.parquet",
)
print(f"Loaded: {result.object_name}")
```
