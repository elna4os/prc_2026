```python
import os

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
training_files = sorted([x for x in os.listdir("../data") if "training" in x])
training_files
```

# Read training data

```python
dfs = []
for file in training_files:
    month = int(file.split("_")[1].split("-")[1])
    df = pd.read_parquet(f"../data/{file}")
    df["month"] = month
    dfs.append(df)
df = pd.concat(dfs, axis=0)
with pd.option_context('display.max_columns', None):
    display(df.head())
```

### TAXITIME_SEC_mvt distribution

```python
df_dep["TAXITIME_SEC_mvt"].describe(percentiles=[0.01, 0.05, 0.25, 0.5, 0.75, 0.95, 0.99])
```

We can notice movements with negative taxi-out time, as well as movements with large taxi-out time (like 1.3e5, which corresponds to around 36 hours). Values between 1 and 99 percentile are in interval ~[300, 2400] seconds or [5, 40] minutes.

Total number of departures with anomalous taxi-out time (outside 1st and 99th percentile) per month:

```python
lower_percentile = df_dep["TAXITIME_SEC_mvt"].quantile(0.01)
upper_percentile = df_dep["TAXITIME_SEC_mvt"].quantile(0.99)
anomalous_movements = df_dep[(df_dep["TAXITIME_SEC_mvt"] < lower_percentile) | (df_dep["TAXITIME_SEC_mvt"] > upper_percentile)]
anomalous_movements.groupby("month").size()
```

Total number of departures with taxi-out time smaller than 1st percentile per month:

```python
anomalous_movements[anomalous_movements["TAXITIME_SEC_mvt"] <= lower_percentile].groupby("month").size()
```

Total number of departures with taxi-out time greater than 99th percentile per month:

```python
anomalous_movements[anomalous_movements["TAXITIME_SEC_mvt"] > upper_percentile].groupby("month").size()
```

### BLOCK_TIME_UTC_mvt and LOBT_flt difference (in minutes)

```python
mask = ~df_dep["BLOCK_TIME_UTC_mvt"].isna() & ~df_dep["LOBT_flt"].isna()
((df_dep[mask]["BLOCK_TIME_UTC_mvt"] - df_dep[mask]["LOBT_flt"]).dt.total_seconds() / 60).plot(
    kind="hist",
    bins=100,
    title="BLOCK_TIME_UTC_mvt and LOBT_flt difference in minutes (for departures)"
)
plt.grid()
```

(for departures where BLOCK_TIME_UTC_mvt and LOBT_flt are both defined):
- As we can see, actual off-block time can be significantly different from the last known off-block time from NM (Network Manager)
- Early pushback, late pushback, and pushback on time are all possible
- The difference can vary from -60 minutes to +60 minutes, with the majority of movements having a difference of less than 10 minutes
- The distribution is slightly skewed to the right, indicating that there are more movements with a positive difference (actual off-block time later than last known off-block time) than movements with a negative difference (actual off-block time earlier than last known off-block time)

Cumulative distribution of the difference between BLOCK_TIME_UTC_mvt and LOBT_flt in minutes (for departures where both are defined):

```python
((df_dep[mask]["BLOCK_TIME_UTC_mvt"] - df_dep[mask]["LOBT_flt"]).dt.total_seconds() / 60).hist(
    cumulative=True,
    bins=100,
    density=True
)
```
