```python
import os

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
training_files = sorted([x for x in os.listdir("../data") if "training" in x])
training_files
```

# Read training data

```python
dfs = []
for file in training_files:
    month = int(file.split("_")[1].split("-")[1])
    df = pd.read_parquet(f"../data/{file}")
    df["month"] = month
    dfs.append(df)
df = pd.concat(dfs, axis=0)
with pd.option_context('display.max_columns', None):
    display(df.head())
```

### Taxi-out time percentiles

```python
df_dep["TAXITIME_SEC_mvt"].describe(percentiles=[0.01, 0.05, 0.25, 0.5, 0.75, 0.95, 0.99])
```

We can notice movements with negative taxi-out time, as well as movements with very large taxi-out time (like 1.3e5, which corresponds to ~36 hours). Values between 1 and 99 percentile are in interval ~[300, 2400] seconds or [5, 40] minutes, which sounds reasonable for most departures.

Total number of departures with anomalous taxi-out time (outside 1st and 99th percentile) per month:

```python
q_1 = df_dep["TAXITIME_SEC_mvt"].quantile(0.01)
q_99 = df_dep["TAXITIME_SEC_mvt"].quantile(0.99)
df_oor = df_dep[(df_dep["TAXITIME_SEC_mvt"] < q_1) | (df_dep["TAXITIME_SEC_mvt"] > q_99)]
df_oor.groupby("month").size()
```

Total number of departures with taxi-out time smaller than 1st percentile per month:

```python
df_oor[df_oor["TAXITIME_SEC_mvt"] <= q_1].groupby("month").size()
```

Total number of departures with taxi-out time greater than 99th percentile per month:

```python
df_oor[df_oor["TAXITIME_SEC_mvt"] > q_99].groupby("month").size()
```

### Median taxi-out time by number of runways

```python
taxi_by_runway_count = (
    df_dep.assign(
        n_runways=df_dep.groupby(["month", "ADEP_mvt"])["RUNWAY_mvt"]
        .transform("nunique")
    )
    .groupby("n_runways", as_index=False)["TAXITIME_SEC_mvt"]
    .median()
)
fig, ax = plt.subplots(figsize=(9, 5))
sns.lineplot(
    data=taxi_by_runway_count,
    x="n_runways",
    y="TAXITIME_SEC_mvt",
    marker="o",
    errorbar=None,
    ax=ax,
)

ax.set_xlabel("Number of runways")
ax.set_ylabel("Median taxi-out time, sec.")
ax.set_xticks(taxi_by_runway_count["n_runways"])
ax.grid()
fig.tight_layout()
```

It seems that airports with more runways tend to have less taxi-out time.

# Approximated taxi-out times

### MVT_TIME_UTC_mvt - SCHED_TIME_UTC_mvt

```python
diff_mvt_sched = (df_dep["MVT_TIME_UTC_mvt"] - df_dep["SCHED_TIME_UTC_mvt"]).dt.total_seconds()
diff_mvt_sched.describe(percentiles=[0.01, 0.05, 0.25, 0.5, 0.75, 0.95, 0.99])
```

Distribution of deltas between such approximated taxi-out time and actual taxi-out time:

```python
(df_dep["TAXITIME_SEC_mvt"] - diff_mvt_sched).describe(
    percentiles=[0.01, 0.05, 0.25, 0.5, 0.75, 0.95, 0.99]
)
```

### MVT_TIME_UTC_mvt - LOBT_flt

```python
diff_mvt_lobt = (df_dep["MVT_TIME_UTC_mvt"] - df_dep["LOBT_flt"]).dt.total_seconds()
diff_mvt_lobt.describe(percentiles=[0.01, 0.05, 0.25, 0.5, 0.75, 0.95, 0.99])
```

Again, let's find deltas between this taxi-out time and target:

```python
(df_dep["TAXITIME_SEC_mvt"] - diff_mvt_lobt).describe(
    percentiles=[0.01, 0.05, 0.25, 0.5, 0.75, 0.95, 0.99]
)
```

Probably, LOBT_flt approximates BLOCK_TIME_UTC_mvt better than SCHED_TIME_UTC_mvt.

### MVT_TIME_UTC_mvt - IOBT_flt

```python
diff_mvt_iobt = (df_dep["MVT_TIME_UTC_mvt"] - df_dep["IOBT_flt"]).dt.total_seconds()
diff_mvt_iobt.describe(percentiles=[0.01, 0.05, 0.25, 0.5, 0.75, 0.95, 0.99])
```

Deltas:

```python
(df_dep["TAXITIME_SEC_mvt"] - diff_mvt_iobt).describe(
    percentiles=[0.01, 0.05, 0.25, 0.5, 0.75, 0.95, 0.99]
)
```

Not surprising, because most of the time IOBT and LOBT are the same.
